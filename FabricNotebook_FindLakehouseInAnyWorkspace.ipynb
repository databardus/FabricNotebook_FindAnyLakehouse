# Option 2: Find the items using SemPy and NotebookUtils
## The path we're looking for requires two pieces of info: 
### <u>Workspace name/id</u> and <u>Lakehouse Name</u>
## Example: fullpath = f"abfss://{workspace_name}@onelake.dfs.fabric.microsoft.com/{lakehouse_name}.Lakehouse"

```python
import notebookutils
import sempy.fabric as fabric


#workspace = fabric.list_workspaces(filter=f"name eq '{workspace_name}'")
workspace = fabric.list_workspaces(filter=f"name eq '{workspace_name}'").iloc[0]
display(workspace)

lakehouse = notebookutils.lakehouse.get(lakehouse_name, workspace.Id)
display(lakehouse)
```

```python
df = spark.read.format("delta").load(f"{lakehouse.properties['abfsPath']}/Tables/dbo/dimension_customer")
display(df)
```

```python
#Polished into a function
def getLakehouseUniversal(workspace_name, lakehouse_name):
    workspace = fabric.list_workspaces(filter=f"name eq '{workspace_name}'").iloc[0]
    lakehouse = notebookutils.lakehouse.get(lakehouse_name, workspace.Id)
    return lakehouse
```
